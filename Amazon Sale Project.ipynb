Bar Chart: Top 10 Product Categories

```python
import matplotlib.pyplot as plt

category_col = [col for col in df.columns if 'category' in col.lower()][0]

print("Using column:", category_col)

category_counts = df[category_col].value_counts().head(10)

plt.figure(figsize=(10, 6))
category_counts.plot(kind='bar')

plt.title('Top 10 Amazon Product Categories')
plt.xlabel('Category')
plt.ylabel('Number of Products')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```
